## Data Cleaning

### Cleaning Decision 1: Remove unnecessary index columns

The columns `Unnamed: 0.1` and `Unnamed: 0` were extra index columns created when the dataset was saved as a CSV. They did not contain useful information about the songs, so we removed them.

```python
df = df.drop(columns=['Unnamed: 0.1', 'Unnamed: 0'])
```

```python
df.columns
```

### Cleaning Decision 2: Check for missing values

Before cleaning the data further, we checked each column for missing values to see if anything needed to be removed or filled in.

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

### Cleaning Decision 3: Remove row with missing track information

There was one row missing the artist, album name, and track name. That same row also had a duration of 0 milliseconds, so it did not seem like a valid track. Since it was only one row out of 114,000, we removed it instead of trying to fill in the missing information.

```python
df = df.dropna(subset=['artists', 'album_name', 'track_name'])
```

We checked the dataset for repeated track IDs to see if the same songs appeared more than once. Some songs were listed multiple times because they were placed under different genres, so we did not want to automatically remove every repeated track ID.

```python
df['track_id'].duplicated().sum()
```

```python
df['track_id'].nunique()
```

```python
duplicate_tracks = df[df.duplicated(subset='track_id', keep=False)]

duplicate_tracks[
    ['track_id', 'artists', 'track_name', 'track_genre', 'popularity']
].sort_values('track_id').head(20)
```

### Cleaning Decision 5: Remove exact duplicate rows

We found 450 rows that were completely identical across every column. Since these rows did not add any new information, we removed them.

```python
df = df.drop_duplicates()
```

Some track IDs appeared more than once because the same song was listed under different genres. Since these rows are not completely identical and the genre information may be useful later, we kept them in the dataset.

## Inferential Analysis

We compared the popularity of explicit and non-explicit songs to see whether there is a statistically significant difference between the two groups. Since some songs appear under multiple genres, we used one row per unique track ID for this test so the same song would not be counted multiple times.

```python
unique_tracks = df.drop_duplicates(subset='track_id')

unique_tracks.groupby('explicit')['popularity'].agg(['count', 'mean', 'median'])
```

### Two-Sample t-test

We used a two-sample t-test to compare the average popularity of explicit and non-explicit songs.

- **Null hypothesis:** There is no difference in average popularity between explicit and non-explicit songs.
- **Alternative hypothesis:** There is a difference in average popularity between explicit and non-explicit songs.

```python
from scipy import stats
```

```python
explicit_popularity = unique_tracks[unique_tracks['explicit'] == True]['popularity']
nonexplicit_popularity = unique_tracks[unique_tracks['explicit'] == False]['popularity']
```

```python
t_stat, p_value = stats.ttest_ind(
    explicit_popularity,
    nonexplicit_popularity,
    equal_var=False
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

### t-test Results

The t-test gave a p-value much smaller than 0.05, so we rejected the null hypothesis. This means there is a statistically significant difference in average popularity between explicit and non-explicit songs. Explicit songs had a higher average popularity score of about 36.9 compared to about 32.9 for non-explicit songs. The difference in average popularity was about 4 points on the 0-100 popularity scale.
